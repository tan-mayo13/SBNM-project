```python
# Calculate absolute edge weights
all_weights = sample_weights.abs().values.flatten()

# Calculate shared thresholds
thresholds = {
    "90th percentile": np.percentile(all_weights, 90),
    "95th percentile": np.percentile(all_weights, 95),
    "99th percentile": np.percentile(all_weights, 99)
}

print("Threshold values:")
for name, value in thresholds.items():
    print(f"{name}: {value:.6f}")


# Calculate density separately for every sample
density_data = []

for sample in sample_weights.columns:
    for percentile, threshold in thresholds.items():

        density = (
            sample_weights[sample].abs() >= threshold
        ).mean()

        density_data.append({
            "sample": sample,
            "threshold": percentile,
            "threshold_value": threshold,
            "network_density": density
        })


# Convert results into a DataFrame
sample_density = pd.DataFrame(density_data)

print("\nSample-specific density:")
display(sample_density.head())


# Summarize density across samples
threshold_df = (
    sample_density
    .groupby("threshold")["network_density"]
    .agg(["mean", "std", "min", "max"])
    .reset_index()
)

print("\nThreshold analysis summary:")
display(threshold_df)
```

```python
import os
import matplotlib.pyplot as plt

os.makedirs("../results", exist_ok=True)

# Save threshold summary
threshold_df.to_csv(
    "../results/threshold_analysis.csv",
    index=False
)

# Save sample-specific density
sample_density.to_csv(
    "../results/sample_specific_density.csv",
    index=False
)

# Plot density distributions
plt.figure(figsize=(9, 6))

sample_density.boxplot(
    column="network_density",
    by="threshold"
)

plt.title("Sample-Specific Network Density")
plt.suptitle("")
plt.xlabel("Threshold")
plt.ylabel("Network Density")
plt.tight_layout()

plt.savefig(
    "../results/sample_specific_density.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Threshold analysis saved successfully!")
```
